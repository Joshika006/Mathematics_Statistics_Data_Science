# Task 1 – Linear Algebra Implementation 

```python
import numpy as np
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])

matrix_a = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 1]
])

matrix_b = np.array([
    [1, 2, 0],
    [3, 1, 4],
    [2, 5, 1]
])


vector_addition = vector_a + vector_b
print("Vector Addition:", vector_addition)

vector_subtraction = vector_a - vector_b
print("Vector Subtraction:", vector_subtraction)

dot_product = np.dot(vector_a, vector_b)
print("Dot Product:", dot_product)

matrix_addition = matrix_a + matrix_b
print("\nMatrix Addition:")
print(matrix_addition)

matrix_multiplication = np.dot(matrix_a, matrix_b)
print("\nMatrix Multiplication:")
print(matrix_multiplication)

determinant = np.linalg.det(matrix_a)
print("\nDeterminant of Matrix A:", determinant)

if determinant != 0:
    inverse = np.linalg.inv(matrix_a)
    print("\nInverse of Matrix A:")
    print(inverse)
else:
    print("\nMatrix A does not have an inverse.")
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

# Task 2 – Probability Concepts 

```python
import numpy as np
from scipy.stats import binom, norm, uniform
sales_success = [0.65, 0.35]

print("Probability of Successful Sales:", sales_success[0])
print("Probability of Unsuccessful Sales:", sales_success[1])

customer_segments = ["Premium", "Regular"]

segment_purchase_probability = {
    "Premium": 0.80,
    "Regular": 0.50
}

print("\nConditional Probability:")
print("P(Purchase | Premium) =", segment_purchase_probability["Premium"])
print("P(Purchase | Regular) =", segment_purchase_probability["Regular"])

P_A = 0.40
P_B = 0.62
P_B_given_A = 0.80

P_A_given_B = (P_B_given_A * P_A) / P_B

print("\nBayes' Theorem:")
print("Updated Probability P(Premium | Purchase) =", P_A_given_B)

n_trials = 20
success_prob = 0.65

binomial_distribution = binom.pmf(
    np.arange(n_trials + 1),
    n_trials,
    success_prob
)

print("\nBinomial Distribution:")
print(binomial_distribution)
normal_mean = 50
normal_std = 10

normal_distribution = norm.rvs(
    loc=normal_mean,
    scale=normal_std,
    size=100
)

print("\nNormal Distribution:")
print(normal_distribution)

uniform_low = 0
uniform_high = 1

uniform_distribution = uniform.rvs(
    loc=uniform_low,
    scale=uniform_high - uniform_low,
    size=100
)
print("\nUniform Distribution:")
print(uniform_distribution)
```

# Descriptive Statistics

```python
import numpy as np
import pandas as pd
import statistics

sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
              190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

sales = np.array(sales_data)
# 1. Mean
mean = np.mean(sales)
# 2. Median
median = np.median(sales)
# 3. Mode
mode = statistics.multimode(sales_data)
# 4. Range
data_range = np.max(sales) - np.min(sales)
# 5. Variance
variance = np.var(sales)
# 6. Standard Deviation
standard_deviation = np.std(sales)
# 7. Skewness and Kurtosis using Pandas
df = pd.DataFrame({"Sales": sales_data})

skewness = df["Sales"].skew()
kurtosis = df["Sales"].kurt()

print("Descriptive Statistics")
print("----------------------")
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Range:", data_range)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
print("Skewness:", skewness)
print("Kurtosis:", kurtosis)

print("\nInterpretation:")

if abs(skewness) < 0.5:
    print("The dataset is approximately symmetric.")
elif skewness > 0:
    print("The dataset is positively skewed.")
else:
    print("The dataset is negatively skewed.")
```
